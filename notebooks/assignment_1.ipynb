```python
import pandas as pd

df = pd.DataFrame({
    'student_id': [101, 102, 103, 104, 105, 106],
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eva', None],
    'test_score': [85.5, 92.0, np.nan, 78.5, 95.0, 88.5],
    'attempts': [1, 2, 3, 1, 4, 2],
    'passed': [True, True, False, False, True, True]
})

print(df.info())
print(df.describe())
print(df.isna().sum())
```

```python
filtered = df[df['test_score'] > 80].sort_values('test_score', ascending=False)
print("Filtered:\n", filtered)

df['good_performer'] = (df['test_score'] > 80) & (df['attempts'] <= 3)
print("With derived column:\n", df)
```

```python
import matplotlib.pyplot as plt

attempts = [1, 2, 3, 4, 5]
train = [0.62, 0.75, 0.83, 0.88, 0.91]
val = [0.60, 0.72, 0.78, 0.79, 0.78]

plt.figure(figsize=(8, 6))
plt.plot(attempts, train, marker='o', label='Training')
plt.plot(attempts, val, marker='s', label='Validation')
plt.xlabel('Training Attempt')
plt.ylabel('Accuracy')
plt.title('Validation Accuracy Plateaus After Attempt 3')
plt.legend()
plt.grid(True, alpha=0.3)
plt.savefig('../reports/day01_chart.png', dpi=150, bbox_inches='tight')
plt.show()
```
